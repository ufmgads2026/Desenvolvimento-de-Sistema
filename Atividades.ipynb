```python
# Exercício 1.1: Dobro, Triplo e Soma
x = 10
dobro = x * 2
triplo = x * 3
soma = dobro + triplo
print("Número:", x)
print("Dobro:", dobro)
print("Triplo:", triplo)
print("Soma do dobro com o triplo:", soma)
```

```python
# Exercício 1.2: Juntando Palavras e Contando Letras
palavra1 = "Olá"
palavra2 = "Mundo"
# Juntando as palavras com um espaço
frase = palavra1 + " " + palavra2
print("Frase gerada:", frase)
# Contando a quantidade de caracteres
total_letras = len(frase)
print("Total de caracteres:", total_letras)
```
